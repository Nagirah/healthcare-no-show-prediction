#### Investigate the date columns

```python
df[["ScheduledDay", "AppointmentDay"]].head()
```

```python
df[["ScheduledDay", "AppointmentDay"]].info()
```

```python
#convert the date columns
df["ScheduledDay"] = pd.to_datetime(df["ScheduledDay"])
df["AppointmentDay"] = pd.to_datetime(df["AppointmentDay"])
```

```python
df[["ScheduledDay", "AppointmentDay"]].info()
```

```python
print("Scheduled appointments:")
print("Earliest:", df["ScheduledDay"].min())
print("Latest:", df["ScheduledDay"].max())

print("\nAppointment dates:")
print("Earliest:", df["AppointmentDay"].min())
print("Latest:", df["AppointmentDay"].max())
```

Create the WaitingDays feature

```python
# Extract only the date portion
df["ScheduledDate"] = df["ScheduledDay"].dt.normalize()
df["AppointmentDate"] = df["AppointmentDay"].dt.normalize()
```

```python
df["WaitingDays"] = (
    df["AppointmentDate"] - df["ScheduledDate"]
).dt.days
```

```python
df[[
    "ScheduledDay",
    "AppointmentDay",
    "ScheduledDate",
    "AppointmentDate",
    "WaitingDays"
]].head()
```

```python
#Check if there are negative waiting days
print("Negative waiting days:", (df["WaitingDays"] < 0).sum())
```

```python
negative_waits = df[df["WaitingDays"] < 0]

negative_waits[[
    "PatientId",
    "ScheduledDay",
    "AppointmentDay",
    "WaitingDays",
    "No-show"
]]
```

```python

# Remove records with negative waiting days
df = df[df["WaitingDays"] >= 0].copy()

print("Dataset shape after removing invalid waiting times:", df.shape)
print("Negative waiting days remaining:", (df["WaitingDays"] < 0).sum())
```

Create useful date features

```python
# Day of the week when the appointment was scheduled
df["ScheduledDayOfWeek"] = df["ScheduledDate"].dt.dayofweek

# Day of the week of the actual appointment
df["AppointmentDayOfWeek"] = df["AppointmentDate"].dt.dayofweek

# Month of the appointment
df["AppointmentMonth"] = df["AppointmentDate"].dt.month
```

Save cleaned notebook

```python
df.to_csv("../data/cleaned_data.csv", index=False)

print("Cleaned dataset saved successfully!")
```
